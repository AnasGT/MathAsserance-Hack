```python
import pandas as pd

# Load your dataset (replace 'your_data.csv' with your actual file)
data = pd.read_excel('data.xlsm')

# Remove leading/trailing spaces from column names
data.columns = data.columns.str.strip()

# Clean categorical columns
categorical_columns = ["Branche", "Code Produit", "Désignation Produit", "Sous-Branche"]

for col in categorical_columns:
    if col in data.columns:
        data[col] = data[col].astype(str).str.strip()

# Ensure "Exercice" is an integer type
if "Exercice" in data.columns:
    data["Exercice"] = pd.to_numeric(data["Exercice"], errors='coerce').astype("Int64")

# Clean numerical columns (remove anomalies)
numeric_columns = data.select_dtypes(include=["float64", "int64"]).columns
for col in numeric_columns:
    data[col] = pd.to_numeric(data[col], errors='coerce')

# Save cleaned dataset
data.to_csv('cleaned_data.csv', index=False)

print("Data cleaning completed. Saved as 'cleaned_data.csv'.")
```

```python
from sklearn.preprocessing import LabelEncoder

# Create a label encoder
le = LabelEncoder()

# Encode 'Désignation Produit' and 'Code Produit'
data['Désignation Produit Encoded'] = le.fit_transform(data['Désignation Produit'])
data['Code Produit Encoded'] = le.fit_transform(data['Code Produit'])

# Calculate the correlation
correlation = data[['Désignation Produit Encoded', 'Code Produit Encoded']].corr().iloc[0, 1]
print(f"Correlation between 'Désignation Produit' and 'Code Produit': {correlation}")
```
